```python
import os
import sys
import shutil
import torch
import psutil
from google.colab import drive

# 1. Mount Google Drive
print("=== 1. Mounting Google Drive ===")
drive.mount('/content/gdrive', force_remount=True)

# Define project directories (using the standard mount point /content/gdrive/MyDrive)
DRIVE_BASE_DIR = "/content/gdrive/MyDrive/VeritasNet"
LOCAL_DIR = "/content/VeritasNet_temp"

# 2. Create Project Directory Structure on Drive
print("\n=== 2. Creating Project Directory Structure ===")
directories = [
    "data/raw",
    "data/processed",
    "notebooks",
    "src",
    "models",
    "reports",
    "results"
]

for d in directories:
    dir_path = os.path.join(DRIVE_BASE_DIR, d)
    os.makedirs(dir_path, exist_ok=True)
print(f"✅ Directory structure verified at: {DRIVE_BASE_DIR}")

# Create temporary local cache directory to speed up IO
os.makedirs(LOCAL_DIR, exist_ok=True)
print(f"✅ Temporary local scratchpad created at: {LOCAL_DIR}")

# 3. System Hardware Diagnostics
print("\n=== 3. System Hardware Diagnostics ===")

# CPU / System RAM
virtual_mem = psutil.virtual_memory()
total_ram_gb = virtual_mem.total / (1024**3)
print(f"🖥️ System RAM: {total_ram_gb:.2f} GB")

# Python & PyTorch Versions
print(f"🐍 Python Version: {sys.version.split()[0]}")
print(f"🔥 PyTorch Version: {torch.__version__}")

# GPU Check
gpu_available = torch.cuda.is_available()
print(f"🎮 GPU Available: {gpu_available}")

if gpu_available:
    gpu_name = torch.cuda.get_device_name(0)
    gpu_properties = torch.cuda.get_device_properties(0)
    total_gpu_mem_gb = gpu_properties.total_memory / (1024**3)
    print(f"   - GPU Model: {gpu_name}")
    print(f"   - Dedicated GPU Memory: {total_gpu_mem_gb:.2f} GB")
else:
    print("⚠️ WARNING: No GPU detected. Please change your runtime type to GPU:")
    print("   -> Go to Menu: Runtime -> Change runtime type -> Select T4 GPU (or other available GPU).")

# 4. Drive Space Check
print("\n=== 4. Storage Space Diagnostics ===")
total, used, free = shutil.disk_usage("/content/gdrive")
print(f"💾 Google Drive Total Space: {total / (1024**3):.2f} GB")
print(f"💾 Google Drive Free Space: {free / (1024**3):.2f} GB")
```
